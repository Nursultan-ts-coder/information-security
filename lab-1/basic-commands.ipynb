# UNIX terminal commands

## Changing current working directory

Go up one directory

```bash
cd ..
```

Go up two directories

```bash
cd ../..
```

Go into a directory

```bash
cd [directory]
```

Go to root directory

```bash
cd /
```

Go to previous directory

```bash
cd -
```

Go to home directory

```bash
cd ~
```

Absolute path example

```bash
cd /Users/nursultanlukmanov/documents/auca
```

## Go to some existing directory and create a new directory

```bash
cd lab-1
mkdir my_dir
```

```bash
mkdir lab-1/mydir
```

## Create a file

```bash
touch file-1.txt
```

## Grant all permissions to file

```bash
chmod 777 file-1.txt
```

## Read file

```bash
cat small-file.txt
```

```bash
less huge-file.txt
```

## Copy file

```bash
cp source.txt destination.txt
```

```bash
cp huge-file.txt small-file.txt
```

```bash
cp huge-file.txt dir-1/backup/
```

```bash
cp huge-file.txt small-file dir-1/backup/
```

## Copy directories

```bash
cp -r dir-1/backup dir-2/backup
```

## Delete file or folder

```bash
rm dir-2/backup/huge-file.txt
```

```bash
rm -r dir-2/backup
```
