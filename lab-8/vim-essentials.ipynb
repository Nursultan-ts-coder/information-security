# Vim Essential Commands Guide

## Navigation Commands

| Command          | Description               |
| ---------------- | ------------------------- |
| `h`              | Move cursor left          |
| `j`              | Move cursor down          |
| `k`              | Move cursor up            |
| `l`              | Move cursor right         |
| `w`              | Move to next word         |
| `b`              | Move to previous word     |
| `0`              | Move to beginning of line |
| `$`              | Move to end of line       |
| `gg`             | Move to beginning of file |
| `G`              | Move to end of file       |
| `:<line_number>` | Jump to specific line     |


## Editing Commands

| Command  | Description                |
| -------- | -------------------------- |
| `i`      | Insert mode before cursor  |
| `a`      | Insert mode after cursor   |
| `o`      | Open new line below        |
| `O`      | Open new line above        |
| `x`      | Delete character at cursor |
| `dd`     | Delete entire line         |
| `dw`     | Delete word                |
| `u`      | Undo last change           |
| `Ctrl+r` | Redo last change           |
| `yy`     | Copy entire line           |
| `yw`     | Copy word                  |
| `p`      | Paste after cursor         |
| `P`      | Paste before cursor        |


## Search & Replace

| Command          | Description               |
| ---------------- | ------------------------- |
| `/pattern`       | Search forward            |
| `?pattern`       | Search backward           |
| `n`              | Next search result        |
| `N`              | Previous search result    |
| `:%s/old/new/g`  | Replace all occurrences   |
| `:%s/old/new/gc` | Replace with confirmation |


## Visual Mode

| Command  | Description                   |
| -------- | ----------------------------- |
| `v`      | Start visual mode (character) |
| `V`      | Start visual mode (line)      |
| `Ctrl+v` | Start visual mode (block)     |


## File Operations

| Command       | Description         |
| ------------- | ------------------- |
| `:w`          | Save file           |
| `:q`          | Quit vim            |
| `:wq`         | Save and quit       |
| `:q!`         | Quit without saving |
| `:e filename` | Open file           |


## Vim Commands in Shell Script


#!/bin/bash

# Essential Vim Commands Reference

# Navigation

vim +1 file.txt # Open file at line 1
vim +/pattern file.txt # Open file and search for pattern

# Open vim with specific commands

vim -c "set number" file.txt # Open with line numbers enabled

# Execute command on file without interactive mode

vim -es +'%s/old/new/g' +'wq' file.txt # Find and replace, then save

# Using vim as a non-interactive editor

echo "Hello World" | vim - +'%s/World/Vim/g' +'%p'

# Batch processing multiple files

for file in \*.txt; do
vim -es +"set number" +"wq" "$file"
done

# Run vim commands from script

cat << 'EOF' | vim -
10 lines of text
to be edited
by vim through stdin
EOF

# Execute vim command directly

vim -c '%s/old/new/g|wq' file.txt

# Open multiple files in split

vim -O file1.txt file2.txt # Open in vertical split
vim -o file1.txt file2.txt # Open in horizontal split


## Pro Tips

- Press `Esc` to exit insert mode and return to normal mode
- Use `:set number` to show line numbers
- Use `:set autoindent` for automatic indentation
- Combine commands: `d2w` means delete 2 words
