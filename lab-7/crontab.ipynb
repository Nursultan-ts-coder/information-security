1. Go to /home directory and create a new .txt file:


````python
cd /home
touch file.txt
````

**Output:** (No visible output - creates file and changes directory)


2. Create a shell script that writes a new line to a file with the current timestamp.


````python
nano write_line.sh
````

**Output:** (Opens nano editor - no visible output)


And paste this code:


````python
#!/bin/bash
echo "New line at $(date)" >> /home/file.txt
````

**Note:** This is the script content to be pasted into nano


3. Make the script executable:


````python
chmod +x /home/write_line.sh
````

**Output:** (No visible output on success)


4. Now, use cron to run this script every minute:


````python
crontab -e
````

**Output:** (Opens default crontab editor)


And paste this new line and save it:


````python
* * * * * /home/write_line.sh
````

**Note:** This line to be added to crontab (runs every minute)


Now you should see a new line in your file.txt every minute. You can check it:


````python
cat /home/file.txt
````

**Output:** (File grows with new timestamps each minute)

```
New line at Thu Mar  7 10:30:00 UTC 2026
New line at Thu Mar  7 10:31:00 UTC 2026
New line at Thu Mar  7 10:32:00 UTC 2026
New line at Thu Mar  7 10:33:00 UTC 2026
```
