### Task 3

Create a script to count word occurrences in a file

Write a shell script named count_word.sh that:  
Takes two arguments: a file name and a word to search for.  
Counts and displays how many times the specified word appears in the file.


```bash
 sudo nano /usr/local/bin/word_count.sh
```

Enter the following script


```bash
filename="$1"
if [[ -f "$filename" ]]; then
  echo "File exists"
else
  echo "File not found: $filename"
  exit 1
fi

awk '{ for (i=1; i<=NF; i++) count[$i]++ }
END { for (w in count) print w, count[w] }' "$filename"
```

Make it executable


```bash
sudo chmod +x /usr/local/bin/word_count.sh
```
