### Task 4

Create a Script to find and delete empty files

Write a shell script named delete_empty_files.sh that:  
Takes a directory as an argument.  
Finds all empty files in that directory and deletes them.  
The script should print the names of deleted files.


````bash
 sudo nano /usr/local/bin/delete_empty_files.sh
````

**Output:** (No visible output - opens nano editor)


````bash
directory="$1"

[[ -d "$directory" ]] || {
  echo "❌ $$directory not found"
  exit 1
}

removeFileNames=()

for file in "$directory"/*; do
    if [[ ! -s "$file" ]]; then
    rm "$file"
    removeFileNames+=("$file")
    fi
done

for file in "${removeFileNames[@]}"; do
    echo "Removed empty file: $file"
done

exit 0
````

**Note:** This is the script content (no direct output)


- Make it executable


````bash
sudo chmod +x /usr/local/bin/delete_empty_files.sh
````

**Output:** (No output on success)


- Run command:


````bash
delete_empty_files.sh test_dir
````

Removed empty file: test_dir/text.txt
