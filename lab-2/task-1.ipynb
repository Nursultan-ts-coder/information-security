### Task 1

Create a script to display  
Write a shell script named greet.sh that:  
Outputs a greeting message, e.g., "Hello, [username]!"


- open nano editor

```bash
nano /usr/local/bin/greet
```

- enter the script:

```bash
echo "Hello, $(whoami)!"
```

- Make it executable by giving the permissions

```bash
sudo chmod +x /usr/local/bin/greet
```
