### Task 2

Create a script that displays time remaining before work day ends

Write a shell script named current_time.sh that:  
Displays the current time and time remaining before work day ends.  
E.g. Current time: 13:30. Work day ends after 4 hours and 30 minutes.


- Create a file using `nano`


```bash
 sudo nano /usr/local/bin/current_time.sh
```

- End enter the following code:


```bash

now=$(date +%s)
work_end_hour=18

# Get today's date and construct end time (portable)
today=$(date +%Y-%m-%d)
end=$(date -d "$today $work_end_hour:00:00" +%s 2>/dev/null || date -j -f "%Y-%m-%d %H:%M:%S" "$today $work_end_hour:00:00" +%s)

remaining=$(( end - now ))

hours=$(( remaining / 3600 ))
minutes=$(( (remaining % 3600) / 60 ))

echo "Time remaining: ${hours}h ${minutes}m"
```

- And make it executable


```bash
sudo chmod +x /usr/local/bin/current_time.sh
```
