1. Open the terminal and run the following commands:  


```bash

sudo adduser hacker
```

The system will prompt you to create and confirm a password and other information for the new user.

2. Verify the user creation by running:

```bash
cat /etc/passwd
```

3. Login as your new user 

```bash
su - hacker
```

4. Verify login by running the following command:

```bash
whoami
```


5. To delete this user you should first kill current login session


```bash
pkill -KILL -u hacker
```

6. Delete the user by command:

```bash
deluser --remove-home --force user1
```

7. Verify deletion:

```bash
cat /etc/passwd
```
