# Lab 6: Permissions in Linux

Permissions control who can access, modify, or execute files and directories.

### Permission Types:

- **Read (r)**: View file contents or list directory contents
- **Write (w)**: Modify file contents or create/delete files in directory
- **Execute (x)**: Run file as program or enter directory

### Permission Format: `-rwxr-xr--`

| Position   | Meaning                                          |
| ---------- | ------------------------------------------------ |
| 1st char   | File type (`-` file, `d` directory, `l` symlink) |
| 2-4 (rwx)  | Owner permissions                                |
| 5-7 (r-x)  | Group permissions                                |
| 8-10 (r--) | Others permissions                               |

### Numeric Values:

- r = 4, w = 2, x = 1
- Sum for each category (e.g., rwx = 7, r-x = 5, r-- = 4)


## 1. Creating a File and Viewing Permissions


````python
touch file.txt
ls -la file.txt
````

**Output:**

```
-rw-r--r-- 1 hacker hackers 0 Mar  7 10:30 file.txt
```


**Example output:**

```
-rw-r--r-- 1 hacker hackers 0 Mar  2 12:12 file.txt
```

**Breakdown:**

- `-rw-r--r--`: File permissions (owner: rw-, group: r--, others: r--)
- `hacker`: File owner
- `hackers`: Group
- `0`: File size in bytes
- `Mar  2 12:12`: Last modified date
- `file.txt`: File name


## 2. Changing Permissions with chmod (Numeric Notation)


````python
# Set rwx for owner (7), r-x for group (5), r-- for others (4)
chmod 754 file.txt
ls -la file.txt
````

**Output:**

```
-rwxr-xr-- 1 hacker hackers 0 Mar  7 10:30 file.txt
```


**Result:** `-rwxr-xr--`

- Owner: rwx (read, write, execute) = 7
- Group: r-x (read, execute) = 5
- Others: r-- (read only) = 4


## 3. Permission Denied Cases

### Case 1: No permissions (000) - Cannot read


````python
# Remove all permissions
chmod 000 file.txt
ls -la file.txt

# Try to read - should fail with "Permission denied"
cat file.txt
````

**Output:**

```
---------- 1 hacker hackers 0 Mar  7 10:30 file.txt
cat: file.txt: Permission denied
```


**Error:** `cat: file.txt: Permission denied`

### Case 2: Read-only (444) - Cannot write


````python
# Set read-only for all
chmod 444 file.txt
ls -la file.txt

# Try to write - should fail with "Permission denied"
echo "test" >> file.txt
````

**Output:**

```
-r--r--r-- 1 hacker hackers 0 Mar  7 10:30 file.txt
bash: file.txt: Permission denied
```


**Error:** `bash: file.txt: Permission denied`

## 4. Changing File Owner and Group

### Commands:

- `sudo chown [owner] file` - Change file owner (requires sudo)
- `sudo chgrp [group] file` - Change file group
- `sudo chown [owner]:[group] file` - Change both at once


````python
# Create a new file
touch file-1.txt
ls -la file-1.txt

# Change group only (to hackers group)
sudo chgrp hackers file-1.txt
ls -la file-1.txt

# Change owner only (to hacker user)
sudo chown hacker file-1.txt
ls -la file-1.txt

# Change both owner and group at once
sudo chown hacker:hackers file-1.txt
ls -la file-1.txt
````

**Output:**

```
-rw-r--r-- 1 hacker hackers 0 Mar  7 10:30 file-1.txt
-rw-r--r-- 1 hacker hackers 0 Mar  7 10:30 file-1.txt
-rw-r--r-- 1 hacker hackers 0 Mar  7 10:30 file-1.txt
```


### How Owner/Group Changes Affect File Usage:

1. **Owner change (`chown`)**:
   - New owner gains owner-level permissions (first 3 permission bits)
   - Previous owner loses owner permissions, falls back to group/others permissions
   - Requires `sudo` to change owner to another user

2. **Group change (`chgrp`)**:
   - Users in new group gain group-level permissions (middle 3 bits)
   - Users in old group lose group permissions, fall back to others permissions

**Example with hacker:hackers:**

```bash
# Original file owned by root
-rw-r--r-- 1 root root 0 Mar  2 12:15 file-1.txt

# After: sudo chown hacker:hackers file-1.txt
-rw-r--r-- 1 hacker hackers 0 Mar  2 12:15 file-1.txt
```

Now:

- User `hacker` has owner permissions (rw-)
- Members of `hackers` group have group permissions (r--)
- Everyone else has others permissions (r--)
- Original owner `root` now only has "others" access (r--)


## 5. Common Permission Values

| Numeric | Symbolic  | Meaning                         |
| ------- | --------- | ------------------------------- |
| 777     | rwxrwxrwx | Full access for everyone        |
| 755     | rwxr-xr-x | Owner full, others read/execute |
| 700     | rwx------ | Owner only                      |
| 644     | rw-r--r-- | Owner read/write, others read   |
| 600     | rw------- | Owner read/write only           |
| 444     | r--r--r-- | Read-only for everyone          |
| 000     | --------- | No access                       |
